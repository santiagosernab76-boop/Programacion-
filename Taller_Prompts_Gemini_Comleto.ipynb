## Problema 1 · Promedio y estado de un estudiante

Un colegio necesita un programa en Python que reciba una lista de calificaciones (0.0 a 5.0) de un estudiante, calcule el promedio y diga si **"Aprueba"** (promedio ≥ 3.0) o **"Reprueba"**.

**Tu tarea:** escribe en `prompt_1` las instrucciones para que Gemini te entregue ese programa.

```python
prompt_1 = """
Crea un programa en Python que permita recibir una lista de calificaciones de un estudiante, teniendo en cuenta que las notas deben estar entre 0.0 y 5.0. El programa debe calcular el promedio de todas las calificaciones y mostrar el resultado. Después, debe indicar si el estudiante "Aprueba" o "Reprueba". Si el promedio es igual o superior a 3.0, debe mostrar "Aprueba"; si es menor a 3.0, debe mostrar "Reprueba".
Quiero que el código sea sencillo y fácil de entender, utilizando listas, ciclos y condicionales básicos. También debe verificar que las notas estén dentro del rango permitido y explicar brevemente cómo funciona el programa.
"""
```

```python
def evaluar_calificaciones(calificaciones):
    for nota in calificaciones:
        if not (0.0 <= nota <= 5.0):
            return f"Error: La nota {nota} está fuera del rango permitido (0.0 - 5.0)."

    if len(calificaciones) == 0:
        return "La lista de calificaciones está vacía."

    promedio = sum(calificaciones) / len(calificaciones)

    if promedio >= 3.0:
        estado = "Aprueba"
    else:
        estado = "Reprueba"

    return f"Promedio: {promedio:.2f} - Estado: {estado}"

notas_estudiante = [3.5, 4.0, 2.8, 4.5]
print(evaluar_calificaciones(notas_estudiante))
```

## Problema 2 · Inventario de una tienda

Una tienda guarda su inventario en un diccionario `{producto: cantidad}`. Necesita un programa que, al recibir ese diccionario y una venta (producto y cantidad), descuente la venta si hay stock suficiente o avise si no lo hay.

**Tu tarea:** escribe en `prompt_2` las instrucciones para que Gemini te entregue ese programa.

```python
prompt_2 = """
Crea un programa en Python para administrar el inventario de una tienda. El inventario debe estar guardado en un diccionario donde cada producto tenga una cantidad disponible. El programa debe permitir registrar una venta indicando el nombre del producto y la cantidad que se quiere vender. Si el producto existe y hay suficiente cantidad disponible, debe descontar las unidades vendidas y mostrar el inventario actualizado. Si no hay suficiente inventario, debe mostrar un mensaje indicando que no hay stock suficiente. Si el producto no existe, también debe avisarlo.
Utiliza diccionarios, condicionales y estructuras básicas de Python. Quiero que el código sea sencillo, organizado y que expliques brevemente cómo funciona.
"""
```

```python
def gestionar_venta(inventario, producto, cantidad_venta):
    if producto not in inventario:
        print(f"El producto '{producto}' no existe en el inventario.")
        return

    if inventario[producto] >= cantidad_venta:
        inventario[producto] -= cantidad_venta
        print(f"Venta exitosa. Inventario actualizado: {inventario}")
    else:
        print(f"No hay stock suficiente. Stock actual de '{producto}': {inventario[producto]}")

tienda_inventario = {"manzanas": 10, "leche": 5, "pan": 2}
print("Inventario inicial:", tienda_inventario)
gestionar_venta(tienda_inventario, "leche", 2)
```

## Problema 3 · Validador de contraseñas

El área de sistemas de una empresa quiere revisar una lista de contraseñas nuevas de empleados. Una contraseña se considera **segura** si tiene al menos 8 caracteres y contiene al menos un número. El programa debe recibir la lista de contraseñas e indicar cuántas son seguras y cuáles no lo son.

**Tu tarea:** escribe en `prompt_3` las instrucciones para que Gemini te entregue ese programa.

```python
prompt_3 = """
Crea un programa en Python que permita revisar una lista de contraseñas de empleados de una empresa. Una contraseña se considera segura si tiene mínimo 8 caracteres y contiene al menos un número. El programa debe recorrer la lista y revisar cada contraseña para determinar si cumple con las dos condiciones. Al finalizar, debe mostrar cuáles contraseñas son seguras, cuáles no lo son y cuántas contraseñas seguras hay en total.
Utiliza listas, ciclos, condicionales y funciones básicas de Python. Quiero que el código sea sencillo y fácil de entender, y que expliques brevemente cómo funciona.
"""
```

```python
def revisar_contrasenias(lista_contrasenias):
    seguras = []
    no_seguras = []

    for pwd in lista_contrasenias:
        largo_suficiente = len(pwd) >= 8
        tiene_numero = any(caracter.isdigit() for caracter in pwd)

        if largo_suficiente and tiene_numero:
            seguras.append(pwd)
        else:
            no_seguras.append(pwd)

    print("Contraseñas seguras:", seguras)
    print("Contraseñas no seguras:", no_seguras)
    print(f"Total de contraseñas seguras: {len(seguras)}")

empleados_contrasenias = ["abc12345", "clave", "segura99", "12345678", "passw"]
revisar_contrasenias(empleados_contrasenias)
```

## Reto final · arma una mini-app solo con prompts

Elige **una** mini-aplicación de consola: calculadora de propinas, gestor de tareas o conversor de unidades. La vas a construir **sin escribir ni una línea de código a mano**, solo con prompts, en dos vueltas:

1. **`prompt_v1`**: pide la primera versión funcional de la app.
2. Prueba el código.
3. **`prompt_v2`**: pídele a Gemini una mejora sobre esa misma app (una validación, una opción nueva, un mensaje más claro). Esto es la **iteración de prompts** que vimos en clase, aplicada a construir software.
4. Prueba la versión mejorada.

```python
prompt_v1 = """
Crea una aplicación de consola en Python que funcione como una calculadora de propinas para un restaurante. El programa debe pedir al usuario el valor total de la cuenta y el porcentaje de propina que desea dejar (10%, 15% o 20%). Después debe calcular cuánto corresponde a la propina y cuál sería el valor total que debe pagar incluyendo la propina. El programa debe mostrar claramente el valor de la cuenta, el valor de la propina y el total a pagar. Utiliza operaciones matemáticas, variables y condicionales básicos. Quiero que sea un código sencillo, fácil de utilizar y que tenga mensajes claros para el usuario.
"""
```

```python
def calcular_propina():
    print("=== CALCULADORA DE PROPINAS ===")
    try:
        cuenta = float(input("Ingrese el valor total de la cuenta: "))
        porcentaje = float(input("Ingrese el porcentaje de propina (10, 15 o 20): "))

        valor_propina = cuenta * (porcentaje / 100)
        total_pagar = cuenta + valor_propina

        print("\n--- RESUMEN DE LA CUENTA ---")
        print(f"Valor de la cuenta: ${cuenta:.2f}")
        print(f"Propina ({porcentaje}%): ${valor_propina:.2f}")
        print(f"Total a pagar: ${total_pagar:.2f}")

    except ValueError:
        print("Error: Por favor, ingrese valores numéricos válidos.")

calcular_propina()
```

```python
prompt_v2 = """
Quiero que mejores la calculadora de propinas que creaste anteriormente. Agrega una validación para que el usuario no pueda ingresar valores negativos o iguales a cero en el valor de la cuenta. También verifica que el porcentaje de propina seleccionado sea únicamente 10%, 15% o 20%. Si el usuario ingresa un dato incorrecto, el programa debe mostrar un mensaje de error y pedir nuevamente la información. Mantén las funciones que ya tenía la primera versión, como calcular la propina y el total a pagar. Conserva el código sencillo y utiliza estructuras básicas de Python. Entrégame el código completo actualizado, no solamente los cambios, y explica brevemente qué mejoras realizaste.
"""
```

```python
def calcular_propina_mejorada():
    print("=== CALCULADORA DE PROPINAS MEJORADA ===")

    while True:
        try:
            cuenta = float(input("Ingrese el valor total de la cuenta (mayor a 0): "))
            if cuenta > 0:
                break
            print("Error: El valor de la cuenta debe ser mayor a cero.")
        except ValueError:
            print("Error: Por favor, ingrese un número válido.")

    while True:
        try:
            porcentaje = float(input("Ingrese el porcentaje de propina (10, 15 o 20): "))
            if porcentaje in [10, 15, 20]:
                break
            print("Error: El porcentaje debe ser exactamente 10, 15 o 20.")
        except ValueError:
            print("Error: Por favor, ingrese un número válido.")

    valor_propina = cuenta * (porcentaje / 100)
    total_pagar = cuenta + valor_propina

    print("\n" + "="*30)
    print("      RESUMEN DE PAGO      ")
    print("="*30)
    print(f"Valor de la cuenta:  ${cuenta:.2f}")
    print(f"Propina ({int(porcentaje)}%):       ${valor_propina:.2f}")
    print(f"Total a pagar:       ${total_pagar:.2f}")
    print("="*30)

calcular_propina_mejorada()
```
